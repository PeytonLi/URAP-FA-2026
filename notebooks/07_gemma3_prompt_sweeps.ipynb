````python
from pathlib import Path
import sys

# Resolve repo root whether the kernel cwd is repo/ or notebooks/
HERE = Path.cwd().resolve()
REPO = HERE if (HERE / "scripts" / "repo_paths.py").exists() else HERE.parent
assert (REPO / "scripts" / "repo_paths.py").exists(), REPO
sys.path.insert(0, str(REPO / "scripts"))
import repo_paths as P
P.ensure_import_paths()
print("REPO =", P.REPO)
````

````python
import pandas as pd
from IPython.display import display, Markdown, Image

sweeps = {
    "prompts_2_4": P.PROMPT_SWEEPS / "prompts_2_4",
    "A": P.PROMPT_SWEEPS / "A",
    "B": P.PROMPT_SWEEPS / "B",
    "C": P.PROMPT_SWEEPS / "C",
    "D": P.PROMPT_SWEEPS / "D",
    "C_vs_prompt_1": P.PROMPT_SWEEPS / "C_vs_prompt_1",
    "D_vs_prompt_1": P.PROMPT_SWEEPS / "D_vs_prompt_1",
}
for name, path in sweeps.items():
    display(Markdown(f"## {name}"))
    if not path.exists():
        print("missing", path)
        continue
    table = path / "statistics_table.csv"
    if table.exists():
        df = pd.read_csv(table)
        fam = df[df["Family"].astype(str).str.contains("primary|vs_prompt_1|behavioral_probe_primary", regex=True)]
        display(fam.head(20))
    # show a couple word-count graphs if present
    pngs = sorted(path.rglob("word_counts_base.png"))[:2]
    for png in pngs:
        display(Markdown(f"`{png.relative_to(P.REPO)}`"))
        display(Image(filename=str(png)))
````
